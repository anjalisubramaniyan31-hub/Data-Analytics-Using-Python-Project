```
total_sales = df["sales_amount"].sum()
print("Total Sales:", total_sales)
```

```
product_sales = df.groupby("region")["sales_amount"].sum()
print(product_sales)
```

```
product_sales.plot(kind='bar')
plt.title("product sales analysis")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```
