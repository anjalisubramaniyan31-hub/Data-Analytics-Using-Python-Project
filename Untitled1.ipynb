```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Dataset Generation
np.random.seed(42)
df = pd.DataFrame({
    'Patient_ID': range(1001, 1201),
    'Age': np.random.randint(5, 85, size=200).astype(float),
    'Disease': np.random.choice(['Flu', 'Diabetes', 'Hypertension', 'Dengue', 'Covid-19', 'Asthma'], size=200),
    'Admission_Date': [pd.to_datetime('2025-01-01') + pd.Timedelta(days=int(np.random.randint(0, 365))) for _ in range(200)]
})
df.loc[df.sample(frac=0.08, random_state=12).index, 'Age'] = np.nan
df.loc[df.sample(frac=0.06, random_state=24).index, 'Disease'] = np.nan

# 2. Data Cleaning (Task 1)
df['Age'] = df['Age'].fillna(df['Age'].median()).astype(int)
df['Disease'] = df['Disease'].fillna('Unknown')

# 3. Data Analysis & Transformation (Task 2 & 3)
df['Age_Group'] = pd.cut(df['Age'], bins=[0, 12, 19, 45, 60, 100], labels=['Child', 'Teenager', 'Adult', 'Middle Aged', 'Senior Citizen'])
df['Admission_Month'] = df['Admission_Date'].dt.strftime('%B')
df['Month_Num'] = df['Admission_Date'].dt.month

monthly_admissions = df.groupby(['Month_Num', 'Admission_Month']).size().reset_index(name='Patient_Count').sort_values('Month_Num')

print("[Disease Count]:\n", df['Disease'].value_counts())
print("\n[Age Group Count]:\n", df['Age_Group'].value_counts())

# 4. Data Visualization (Task 4)
sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
sns.countplot(data=df, x='Disease', order=df['Disease'].value_counts().index, palette='Set2')
plt.title('Patient Count by Disease')
plt.xticks(rotation=30)

plt.subplot(1, 2, 2)
sns.lineplot(data=monthly_admissions, x='Admission_Month', y='Patient_Count', marker='o', color='red', linewidth=2)
plt.title('Monthly Admission Trends')
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```
