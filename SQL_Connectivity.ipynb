```python
import sqlite3
import pandas as pd

# Connect to SQLite database (creates 'school.db' if it doesn't exist)
conn = sqlite3.connect('College.db')
cursor = conn.cursor()

# Create table if it doesn't exist
cursor.execute('''
    CREATE TABLE IF NOT EXISTS students (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        age INTEGER

    )
''')
conn.commit()
print("Database 'College.db' and table 'students' are ready.")
```

```python
import sqlite3
import ipywidgets as widgets
from IPython.display import display, clear_output

# Database Connection
conn = sqlite3.connect("College.db") # Changed from "students.db" to "College.db"
cursor = conn.cursor()

# Input Widget
find_name = widgets.Text(description="Find Name:")
submit_btn = widgets.Button(description="Submit", button_style="success")
output = widgets.Output()

# Search Function
def search_name(b):
    with output:
        clear_output()

        name = find_name.value

        cursor.execute("SELECT id, name, age FROM students WHERE name=?", (name,))
        result = cursor.fetchone()

        if result:
            print("ID   :", result[0])
            print("Name :", result[1])
            print("Age  :", result[2])
        else:
            print("Student Not Found")

submit_btn.on_click(search_name)

display(find_name, submit_btn, output)
```
